```python
from sklearn import svm
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
```

# 1. Loading the Data
# This is the "Big Data" mentioned in your summary. 
# While Iris is small, the process remains the same for massive datasets

# 2. Organizing Features and Labels
# X = Measurements (Features), y = Species (Target)

# 3. Automation: The Train/Test Split
# You are setting 104 samples for testing and the remaining (46) for training.

```python
data = load_iris()
X, y = data.data, data.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=134)
```

```python
clf = svm.SVC(kernel='linear')
clf.fit(X_train,y_train)
accuracy = clf.score(X_train,y_train)
print(f"Model Accuracy:{accuracy*100:.2f}")
```

# 1. Automation: This replaces the manual 'train_test_split' for evaluation.
# It splits the data into 'cv' number of folds (in this case, 5).
# The system learns iteratively on 4 folds and predicts on the 5th, repeating 5 times.

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(clf, X, y, cv=10)

```

# 2. Results: Each score represents the accuracy of one "pass" through the data.
# 3. Acceleration: Quickly generating a stable, averaged metric for big data.

```python
print(f"All scores: {scores}")
print(f"Average stable accuracy: {scores.mean()*100:.2f}%")
```
